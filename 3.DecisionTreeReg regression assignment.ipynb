```python

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python

data = pd.read_csv("insurance_pre.csv")
```

```python
data
```

```python
data=pd.get_dummies(data,drop_first=True)
```

```python
data
```

```python
independent = data[["age","sex_male","bmi","children","smoker_yes"]]
dependent = data[["charges"]] 
```

```python

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(independent, dependent, test_size = 1/3, random_state = 0)
```

```python
from sklearn.tree import DecisionTreeRegressor
regressor=DecisionTreeRegressor(criterion='friedman_mse',splitter='best',max_features='log2') #random,best,absolute_error,squared_error 
regressor=regressor.fit(X_train,y_train)#sqrt,log2
```

```python
y_pred=regressor.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
```
