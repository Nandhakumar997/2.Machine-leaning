```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = pd.read_csv("insurance_pre.csv")
```

```python
data
```

```python
data = pd.get_dummies(data,dtype=int, drop_first=True)
data.head()
```

```python
independent = data[["age","sex_male","bmi","children","smoker_yes"]]
dependent = data[["charges"]] 
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import AdaBoostRegressor
param_grid = {'n_estimators':[1,10,100,1000],
             'loss':['linear','square','exponential']}
grid = GridSearchCV(AdaBoostRegressor(random_state=0), param_grid,refit=True, verbose=3,n_jobs=-1,cv=3)
grid.fit(independent,dependent)
```

```python
re=grid.cv_results_
grid_predictions = grid.predict(independent)
```

```python
from sklearn.metrics import r2_score
r_score=r2_score(dependent,grid_predictions)
print("The R_score value for best parameter {}:".format(grid.best_params_),r_score)
```

```python
table=pd.DataFrame.from_dict(re)
```
